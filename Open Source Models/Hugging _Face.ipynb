#### Pipelines:
Using Pipelines from Hugging Face
A simple way to run inference for common tasks, without worrying about all the plumbing, picking reasonable defaults.

How it works:
STEP 1: Create a pipeline - a function you can then call

my_pipeline = pipeline(task, model=xx, device=xx)
If you don't specify a model, then Hugging Face picks one for you that's the default for the task. Specify "cuda" for the device to use an NVIDIA GPU like the one on the T4. Specify "mps" on a Mac.

STEP 2: Then call it as many times as you want:

my_pipeline(input1)
my_pipeline(input2)

https://huggingface.co/docs/transformers/main_classes/pipelines
https://huggingface.co/tasks

```python
# Example of using the Task "sentimet analyzer" to evaluate the sentence :

my_simple_sentiment_analyzer = pipeline("sentiment-analysis", device="cuda")
result = my_simple_sentiment_analyzer("I'm super excited to be on the way to LLM mastery!")
print(result)

```

#### Tokenizers:
We use token IDs for tokens ( words or word fragments ) in AI models because all of models are based on statistical models, which only recognize and take numbers, so we had to use the numbers to represent words, pixels ....etc in order to feed them to the statistical models.
We will see the differences between the type of models based on the output of these models, the output of the Llama 3.1 8B is similar to the output of most "Base" models, while "Chat" and "Reasoning" models expect things like "system prompt" and "user prompt". Although some base modesl have chat variants.

https://colab.research.google.com/drive/1WD6Y2N7ctQi1X9wa6rpkg8UfyA4iSVuz?usp=sharing

```python
# The code is run on Colab.

from google.colab import userdata
from huggingface_hub import login
from transformers import AutoTokenizer
```

```python


tokenizer = AutoTokenizer.from_pretrained('meta-llama/Meta-Llama-3.1-8B', trust_remote_code=True)

text = "I am excited to show Tokenizers in action to my LLM engineers"
tokens = tokenizer.encode(text)
tokens

 # result outputed on Colab:

[128000,
 40,
 1097,
 12304,
 311,
 1501,
 9857,
 12509,
 304,
 1957,
 311,
 856,
 445,
 11237,
 25175]

# We can go from tokens to text by doing the reverse function of encode: decode:

text = tokenizer.decode(tokens)

# Colab output of Llama 3.1:

'<|begin_of_text|>I am excited to show Tokenizers in action to my LLM engineers'
```
