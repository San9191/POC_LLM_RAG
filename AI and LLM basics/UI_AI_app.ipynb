```python
from openai import OpenAI
from dotenv import load_dotenv
import gradio as gr
load_dotenv()
openai = OpenAI()
```

```python
# Basic Gradio app
# launch(inbrowser=true, auth=("MyName", "pass")) to open in browser &  for authentication.

def invert_word(text):
  return text[::-1]

gr.Interface(fn=invert_word, inputs="textbox", outputs="textbox").launch()
```

```python
system_message = "You are a helpful assistant"

def message_gpt(prompt):
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
    return response.choices[0].message.content

```

```python
user_input = gr.Textbox(label="Your message:" , info="Enter a Question", lines=7)
system_output = gr.Textbox(label="Response:", lines=8)

view=gr.Interface(
  fn=message_gpt,
  title="GPT",
  inputs=[user_input],
  outputs=[system_output],
  examples=["Hello", "Cat"],
  flagging_mode="never"
)
view.launch()
```
