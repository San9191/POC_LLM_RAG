```python
from openai import OpenAI
from dotenv import load_dotenv
import gradio as gr
import json
from IPython.display import Markdown, display

load_dotenv()
openai = OpenAI()
```

```python
MODEL='gpt-4o-mini'
system_message="You are a sales a associate at a shop, you want the customers to know that we have a sale of 50 per cent on shoes "
```

```python
# NOTE : Gradio can track the history when gr.ChatInterface or gr.Chatbot are used and passes it into
# the called function if it has hitory as a parameter.

def chat1(message, history):
  return f"You said {message}, and history is {history}, but I stil say What?!!"
```

```python
gr.ChatInterface(fn=chat1).launch()
```

```python
def chat(message, history):
  history = [{"role":h["role"], "content":h["content"]} for h in history]
  messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
  response = openai.chat.completions.create(model=MODEL, messages=messages)
  return response.choices[0].message.content
```

```python
gr.ChatInterface(fn=chat,).launch()
```
